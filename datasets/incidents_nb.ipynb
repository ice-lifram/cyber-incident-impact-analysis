# **Incidents Master Analysis**

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('incidents_master.csv')

incident_id = df['incident_id']
revenue = df['company_revenue_usd']
attack_vector = df['attack_vector_primary']
downtime_hours = df['downtime_hours']
```

### **Descriptive Analysis**

--------

**Information of Dataset**

```python
print(df.info())
```

- This contains all the initial information about the incidents masters dataset
- Company details, attack chaines, disclosure dates, and quality grades are included within this dataset

--------

**Describing all numerical data points**

```python
print(df.describe())
```

- This describes the descriptive statistics of all numerical data included within the dataset.
- With this function includes:
    - count : the number of valid/non-null data points; assess the completeness of the dataset by showing how many entries are valid.
    - mean : the average
    - std : the standard deviation
    - min : the minimum/lowest value 
    - 25%. 50%, 75% : the quartiles (Q1, Q2 (median), Q3 respectively)
    - max : the maximum/highest value

**Number of Attack Vectors**

```python
# Number of Attack Vectors
attack_vectors_numbers = attack_vector.value_counts()
print(attack_vectors_numbers)
```

**Downtime Hours**

```python
# Average Downtime Hours
print(downtime_hours.describe())
print(downtime_hours.skew())
```

- This data describes the central tendency; mean, median, and standard deviation is shown.

**Number of Quality Grades betwwen Companies**

**Attack Revenue + Downtime Analysis**

```python
# Attack Vector + Downtime Analysis
attack_downtime = df.groupby("attack_vector_primary")["downtime_hours"].agg([
    "count",
    "median",
    "mean",
    ])

print(attack_downtime)
```

- This section analyzes the total count, median, and 
mean of downtime hours of each attack.
- The NaN value implies an unobserved downtime hours.

---

**Attack Vectors + Data Compromised Analysis**

```python
# Attack Vector + Data Compromised Records Analysis
print("\nAttack Vector + Data Compromised Analysis")
record_compromised = df.groupby("attack_vector_primary")["data_compromised_records"].agg([
    "count",
    "mean",
    "median",
    "skew"
])

print(record_compromised)
```

- This section provides the count, mean, median, and skew of data compromised on their respective attack vectors. 
- The NaN implies unobserved records.

---

**Industry Sectors affected**

```python
# Industries affected
industries = df['industry_primary'].value_counts()
print(industries)
industries_downtime = df.groupby('industry_primary')['downtime_hours'].agg([
    "count",
    "mean",
    "median"
])
print(industries_downtime)

print()

industries_records = df.groupby("industry_primary")['data_compromised_records'].agg([
    "count",
    "mean",
    "median"
])
print(industries_records)
```

**Number of Missing Values**

```python
print("\nMissing Values")
print("--------------------")

missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(2)

missing_df = pd.DataFrame({
    "Missing Count": missing,
    "Missing %": missing_pct
})

print(missing_df.sort_values("Missing %", ascending=False))
```

- This output checks all missing values between columns, showing the number of cells within the spreadsheet that contains no value.

### **Graphs**

---

**Number of Attack Vectors**

```python
# Number of Attack Vectors
attack_vectors_numbers = attack_vector.value_counts()

plt.figure(figsize=(12,8))
attack_vector_plots = attack_vectors_numbers.plot(kind='bar')
plt.title("Number of Attack Vector")
```

**Number of Public and Private Companies**

```python
# Public/Private Companies
company_type = (
    df["is_public_company"]
    .replace({True: "Public", False: "Private"})
    .value_counts()
)
plt.figure(figsize=(12,6))
company_type.plot(kind="bar")

plt.title("Public vs Private Companies")
plt.xlabel("Company Type")
plt.ylabel("Number of Incidents")
plt.show()
```

**Records of Data Compromised**

```python
# Data Compromised Records
data_com_rec = df['data_compromised_records']
plt.figure(figsize=(10,6))

df["data_compromised_records"].dropna().plot(kind='box')

plt.title("Records Compromised Distribution")
plt.ylabel("Records")

plt.show()
```

**Downtime Hours by Attack Vectors**

```python
attack_downtime = (
    df.groupby("attack_vector_primary")["downtime_hours"]
      .median()
      .sort_values(ascending=False)
)

plt.figure(figsize=(10,6))

attack_downtime.plot(kind="bar")

plt.title("Median Downtime Hours by Attack Vector")
plt.xlabel("Attack Vector")
plt.ylabel("Median Downtime (Hours)")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

Note that backdoor, data breach, phishing, and trojan have undocumented downtime hours

```python
plt.figure(figsize=(12,6))

df.boxplot(
    column="downtime_hours",
    by="attack_vector_primary",
    rot=45
)

plt.title("Downtime Hours by Attack Vector")
plt.suptitle("")
plt.xlabel("Attack Vector")
plt.ylabel("Downtime Hours")

plt.tight_layout()
plt.show()
```

```python
records_by_attack = (
    df.groupby("attack_vector_primary")["data_compromised_records"]
      .agg(["count", "mean", "median"])
      .sort_values("median", ascending=False)
)

records_median = (
    df.groupby("attack_vector_primary")["data_compromised_records"]
      .median()
      .sort_values(ascending=False)
)

plt.figure(figsize=(10,6))

records_median.plot(kind="bar")

plt.title("Median Records Compromised by Attack Vector")
plt.xlabel("Attack Vector")
plt.ylabel("Median Records Compromised")

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
industry_median = (
    df.groupby("industry_primary")["downtime_hours"]
      .median()
      .sort_values(ascending=False)
)

plt.figure(figsize=(10,6))

industry_median.plot(kind="bar")

plt.title("Median Downtime Hours by Industry")
plt.xlabel("Industry")
plt.ylabel("Median Downtime (Hours)")

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```
