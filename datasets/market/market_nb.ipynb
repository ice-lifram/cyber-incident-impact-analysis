# **Market Impact Analysis**

```python
import pandas as pd

df = pd.read_csv('market_impact.csv')
```

```python
print(df.info())
```

**Analysis**

```python
# Market reaction summary
market_reaction = df[[
    "abnormal_return_1d",
    "abnormal_return_7d",
    "abnormal_return_30d",
    "car_neg1_to_pos1",
    "car_0_to_7",
    'car_0_to_30',
    "car_0_to_90"
]]

print(market_reaction.describe())
```

```python
# Volatility Summary
volatility_summary = df[[
    "pre_incident_volatility_30d",
    "post_incident_volatility_30d"
]]

print(volatility_summary.describe())
```
